```python
# Install required libraries
!pip install diffusers accelerate transformers datasets

# Import necessary modules
from datasets import load_dataset, Dataset, Image
import json, os
import pandas as pd
import torch
from matplotlib.pyplot import imshow

# Disable Weights & Biases logging (remove if you want to log)
os.environ["WANDB_MODE"] = "offline"

# Clone latest diffusers and install
!git clone https://github.com/huggingface/diffusers
!cd diffusers && pip install -e .

# Check if train_dreambooth_lora_sd3.py exists, if not, we'll need to create it
!find /content/diffusers/examples -name "*.py" | grep -E "dreambooth|sd3"

# Create train_dreambooth_lora_sd3.py file if it doesn't exist
# We'll adapt this from the existing dreambooth scripts
!cp /content/diffusers/examples/dreambooth/train_dreambooth_lora.py /content/diffusers/examples/dreambooth/train_dreambooth_lora_sd3.py

# Let's load the dataset
ds_ = load_dataset("SimulaMet-HOST/Kvasir-VQA")['raw']
captions = pd.read_json("https://raw.githubusercontent.com/simula/ImageCLEFmed-MEDVQA-GI-2025/refs/heads/main/kvasir-captions.json").to_dict()

# Save images locally
img_path = "kvasir-vqa/images"
os.makedirs(img_path, exist_ok=True)
existing_files = set(os.listdir(img_path))

ds = ds_.map(lambda e: {
    "captions": (
        e['image'].save(f"{img_path}/{e['img_id']}.jpg")
        or existing_files.add(f"{e['img_id']}.jpg")
        if f"{e['img_id']}.jpg" not in existing_files else None
    ) or f"{img_path}/{e['img_id']}.jpg",
})

print(ds[0])  # Preview one sample

# Authenticate with HuggingFace
!git config --global credential.helper manager
!huggingface-cli login
```

```python
# Prepare the dataset in a format suitable for dreambooth
datax = {
    "image": [],
    "text": []  # Using "text" as the column name to match the example
}

for img, captions_group in captions.items():
    for caption in captions_group.values():
        for cap in caption:
            # Enhanced prompt formatting with a consistent class name
            # For DreamBooth, we want to use a consistent class identifier
            enhanced_cap = f"A clinical colonoscopy image of {cap}. High-definition medical photograph with clear visibility of mucosal details and tissue characteristics."
            datax["image"].append({"path": f"{img_path}/{img}.jpg"})
            datax["text"].append(enhanced_cap)

dataset = Dataset.from_dict(datax).cast_column("image", Image())
dataset = dataset.shuffle(seed=42)  # Set seed for reproducibility

# Take a subset for faster training
subset_size = int(len(dataset) * 0.1)  # 10% for demo
dataset_subset = dataset.select(range(subset_size))

# Push to HuggingFace Hub
username = "sageofai"  # Replace with your username
dataset_name = f"{username}/kvasir_dreambooth"
dataset_subset.push_to_hub(dataset_name)

# Create a simple instance prompt that summarizes the style we want
instance_prompt = "Endoscopic image showing a 12mm sessile polyp in the descending colon."
```
